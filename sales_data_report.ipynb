```python
import numpy as np

np.random.seed(42)

months = np.arange(1, 13)

stores = np.arange(1, 11)

sales = np.random.randint(10000, 100001, size=(10, 12))

print("Months:", months)
print("\nSales Data Shape:", sales.shape)
print("\nSales Data:\n", sales)
```

```python
q1 = sales[:, 0:3]
q2 = sales[:, 3:6]
q3 = sales[:, 6:9]
q4 = sales[:, 9:12]

print("Q1 Shape:", q1.shape)
print("Q2 Shape:", q2.shape)
print("Q3 Shape:", q3.shape)
print("Q4 Shape:", q4.shape)

store1 = sales[0, :]
print("\nStore 1 Sales:", store1)

store3_q2 = sales[2, 3:6]
print("\nStore 3 Q2 Sales:", store3_q2)
```

```python
target = 50000

high_sales = sales > target

print("Sales above ₹50,000:\n", sales[high_sales])

store_index, month_index = np.where(sales > target)

print("\nStore numbers:", store_index + 1)
print("Month numbers:", month_index + 1)
```

```python
condition1 = sales > 50000

condition2 = sales < 80000

result_and = sales[condition1 & condition2]

result_or = sales[condition1 | (sales < 20000)]

result_not = sales[~condition1]

print("AND result:", result_and)
print("OR result:", result_or)
print("NOT result:", result_not)
```

```python
store_totals = np.sum(sales, axis=1)

average_sales = np.mean(store_totals)

below_average = store_totals < average_sales

print("Average yearly sales:", average_sales)
print("Stores below average:", stores[below_average])
print("Their yearly sales:", store_totals[below_average])
```

```python
selected_stores = sales[[1, 4, 7], :]

print("Selected store sales:\n", selected_stores)

monthly_totals = np.sum(sales, axis=0)

best_month_indices = np.argsort(monthly_totals)[-3:][::-1]

print("\nBest month numbers:", best_month_indices + 1)
print("Sales in best months:", monthly_totals[best_month_indices])

best_month_sales = sales[np.ix_([1, 4, 7], best_month_indices)]

print("\nSelected stores in best months:\n", best_month_sales)
```

```python

growth_targets = np.linspace(0.05, 0.15, 10)

target_sales = store_totals * (1 + growth_targets)

growth_amount = target_sales - store_totals

print("Store | Current Sales | Growth % | Target Sales")

for i in range(10):
    print(
        f"{stores[i]:5} | "
        f"₹{store_totals[i]:12,.0f} | "
        f"{growth_targets[i] * 100:7.2f}% | "
        f"₹{target_sales[i]:12,.0f}"
    )
```

```python
monthly_totals = np.sum(sales, axis=0)
monthly_average = np.mean(sales, axis=0)

print("MONTHLY SALES REPORT")
print("-" * 45)

for i in range(12):
    print(
        f"Month {months[i]:2}: "
        f"Total = ₹{monthly_totals[i]:10,.0f}, "
        f"Average per store = ₹{monthly_average[i]:,.2f}"
    )
```

```python
quarters = [q1, q2, q3, q4]

print("QUARTERLY SALES REPORT")
print("-" * 40)

for i, quarter in enumerate(quarters, start=1):
    print(f"\nQ{i} Total Sales: ₹{np.sum(quarter):,.0f}")

    quarter_store_totals = np.sum(quarter, axis=1)

    print("Store-wise sales:", quarter_store_totals)

    best_store_index = np.argmax(quarter_store_totals)

    print(
        f"Best performing store: Store {best_store_index + 1}"
    )
```

```python
total_revenue = np.sum(sales)

best_store = np.argmax(store_totals)
lowest_store = np.argmin(store_totals)

best_month = np.argmax(monthly_totals)
lowest_month = np.argmin(monthly_totals)

print("=" * 45)
print("FINAL YEARLY SALES SUMMARY")
print("=" * 45)

print(f"Total Revenue: ₹{total_revenue:,.0f}")
print(f"Average Store Revenue: ₹{np.mean(store_totals):,.2f}")
print(f"Best Store: Store {best_store + 1}")
print(f"Lowest Store: Store {lowest_store + 1}")
print(f"Best Month: Month {best_month + 1}")
print(f"Lowest Month: Month {lowest_month + 1}")
print(f"Stores Below Average: {np.sum(below_average)}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(months, monthly_totals, marker="o")

plt.title("Monthly Sales Across All Stores")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(months)
plt.grid(True)

plt.show()
```

Month Over month Sale

```python
monthly_totals = np.sum(sales, axis=0)

monthly_change = np.diff(monthly_totals)

growth_percentage = (
    monthly_change / monthly_totals[:-1]
) * 100

print("MONTHLY SALES GROWTH REPORT")

for i in range(1, 12):
    print(
        f"Month {i+1}: Change = ₹{monthly_change[i-1]:,.0f}, "
        f"Growth = {growth_percentage[i-1]:.2f}%"
    )
```

Target achievement tracker

```python
targets = np.linspace(500000, 800000, 10)

achievement = (store_totals / targets) * 100

achieved = store_totals >= targets

print("STORE TARGET REPORT")

for i in range(10):
    status = "ACHIEVED" if achieved[i] else "NOT ACHIEVED"

    print(
        f"Store {i+1}: "
        f"Sales ₹{store_totals[i]:,.0f} | "
        f"Target ₹{targets[i]:,.0f} | "
        f"Achievement {achievement[i]:.2f}% | "
        f"{status}"
    )

print("\nTargets achieved:", np.sum(achieved))
print("Targets missed:", np.sum(~achieved))
```
